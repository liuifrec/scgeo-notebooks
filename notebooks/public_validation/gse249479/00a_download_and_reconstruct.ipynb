# Dataset B GSE249479 Phase 1A: public download and sparse reconstruction

This notebook discovers the official public GEO supplementary files for the selected RNA samples, downloads only the processed 10x MEX components, inventories public metadata containers without loading them, and reconstructs a sparse raw-count AnnData object. It does not run normalization, HVG selection, PCA, integration, embedding, RNA velocity, ScGeo, or comparator analysis.

## Reader guide

- **Purpose:** Dataset B GSE249479 Phase 1A: public download and sparse reconstruction in the frozen revision workflow.
- **Inference scope:** Dataset B is descriptive_only because no valid biological-replicate identity was recovered. Cells, conditions, libraries, and inferred partitions are not replicates.
- **Inputs:** Official study metadata and the immutable workstation H5AD, where applicable.
- **Implementation:** `scripts/build_gse249479_sparse_h5ad.py`.
- **Outputs:** Ignored `results/public_validation/gse249479_dataset_b/` artifacts.
- **Frozen findings:** Retain the accepted descriptive TNF/LPS geometry, representation sensitivity, signature, abundance/distribution, and official R Augur evidence. The Python comparator remains supplementary only.
- **Limitations:** No population-level biological inference is available; nested PCA views are not independent confirmations and UMAP is display-only.

This source notebook is intentionally a thin, output-free entry point. The testable implementation is maintained in scripts/build_gse249479_sparse_h5ad.py. Executed review copies and generated artifacts are written under the ignored results directory.


```python
from pathlib import Path
import os
import sys

ROOT = Path.cwd().resolve()
sys.path.insert(0, str(ROOT / "scripts"))

from gse249479_memory_safe import (
    configured_paths,
    ensure_output_tree,
    load_config,
    require_active_branch,
)
from download_gse249479_processed import run as run_download
from build_gse249479_sparse_h5ad import run as run_build

CONFIG = load_config(ROOT)
PATHS = configured_paths(CONFIG, ROOT)
DATA_DIR = PATHS["data_dir"]
OUTPUT_DIR = PATHS["output_dir"]
OUTPUT_H5AD = PATHS["input_h5ad"]
ensure_output_tree(OUTPUT_DIR)
ACTIVE_BRANCH = require_active_branch(ROOT, CONFIG["required_git_branch"])
MEMORY_THRESHOLD_GB = float(os.environ.get(CONFIG["memory_threshold_gb_env"], CONFIG["default_memory_threshold_gb"]))

{
    "active_branch": ACTIVE_BRANCH,
    "data_dir": str(DATA_DIR),
    "output_dir": str(OUTPUT_DIR),
    "output_h5ad": str(OUTPUT_H5AD),
    "memory_threshold_gb": MEMORY_THRESHOLD_GB,
}
```

```python
download_summary = run_download(
    root=ROOT,
    data_dir=DATA_DIR,
    output_dir=OUTPUT_DIR,
    memory_threshold_gb=MEMORY_THRESHOLD_GB,
)
{
    "n_downloaded_files": download_summary["n_downloaded_files"],
    "downloaded_bytes": download_summary["downloaded_bytes"],
    "manifest_path": download_summary["manifest_path"],
    "checksum_path": download_summary["checksum_path"],
    "metadata_inventory_path": download_summary["metadata_inventory_path"],
    "peak_rss_gb": download_summary["peak_rss_gb"],
}
```

```python
build_summary = run_build(
    root=ROOT,
    data_dir=DATA_DIR,
    output_dir=OUTPUT_DIR,
    output_h5ad=OUTPUT_H5AD,
    memory_threshold_gb=MEMORY_THRESHOLD_GB,
)
{
    "output_h5ad": build_summary["output_h5ad"],
    "output_h5ad_size_bytes": build_summary["output_h5ad_size_bytes"],
    "shape": build_summary["shape"],
    "nnz": build_summary["nnz"],
    "sparse_format": build_summary["sparse_format"],
    "matrix_dtype": build_summary["matrix_dtype"],
    "condition_counts": build_summary["condition_counts"],
    "peak_rss_gb": build_summary["peak_rss_gb"],
}
```
