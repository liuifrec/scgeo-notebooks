# 10 · The stored-evidence state response compass

**ScGeo 0.2.0 learning track.** Keep numerical displacement and alignment in the analysis representation while the display embedding positions states. The synthetic sparse state remains visible without pretending that its control shift is supported.

Use the [separate tutorial environment](README.md). Run cells in order in a fresh kernel. Outputs are written to a new ignored results directory; manuscript inputs and outputs are not used.

```python
from pathlib import Path
import sys
import json
import numpy as np
import pandas as pd
import scgeo as sg
from IPython.display import display

cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents)
             if (p / "scripts/scgeo_v020_examples/SOURCE_PART1.json").is_file()), None)
if root is None:
    raise RuntimeError("Open this notebook from the scgeo-notebooks checkout.")
examples = root / "scripts/scgeo_v020_examples"
if str(examples) not in sys.path:
    sys.path.insert(0, str(examples))  # only tutorial helpers, not package source
from tutorial_support import setup, save_preview, show_gallery, FORMATS
out = setup(root, "10_stored_compass")
```

```python
from compass_evidence_gallery import prepare_example, run_gallery, STATES
adata, report, alignment = prepare_example(seed=83)
display(report)
display(alignment)
result = sg.pl.recovery_compass(adata, node_key="state", condition_key="condition",
    group0="control", group1="treated", basis="display", paga_key="paga_fixture",
    graph_node_order=STATES, mode="stored", shift_df=report, alignment_df=alignment,
    analysis_rep="X_pca", effect_metric="delta_norm", node_position="pooled",
    sample_key="sample", min_cells=10, node_size_mode="constant",
    layout="summary", ring_mode="none", max_points=450, return_result=True, show=False)
display(result.data.keys())
save_preview(result, out, "stored_compass")
```

## Concordance, independent availability and rendering controls

Only compatible source tables belong in the same contrast. An unsupported shift is not a near-zero shift, and a missing OOD score is not measured zero burden. Alignment is with/opposed to control-to-treated displacement, not proof of recovery.

```python
result = sg.pl.geometry_velocity_concordance(report, alignment,
    alignment_state_key="node", effect_metric="delta_norm", analysis_rep="X_pca",
    min_cells=10, min_samples=2, sample_count_keys=("n_samples0","n_samples1"),
    highlight_states=["Shared"], max_labels=1, return_result=True, show=False)
save_preview(result, out, "geometry_velocity_concordance")
```

## Run the stored-evidence regression gallery

The gallery checks invariance to changes in display coordinates and vectors, independent evidence availability, missingness, contrast-scoped velocity, sample zeros, threshold selection and display-only subsampling. Timings are optional observations, not machine-independent speed claims.

```python
validation = run_gallery(out / "gallery", formats=FORMATS)
display(validation["checks"])
show_gallery(out / "gallery", limit=3)
```
