# 04 · Local velocity grids, streamlines and matched conditions

**ScGeo 0.2.0 learning track.** Work with constructed projected vectors, not fitted RNA velocity or observed trajectories. The third simulated condition is excluded from the requested contrast.

Use the [separate tutorial environment](README.md). Run cells in order in a fresh kernel. Outputs are written to a new ignored results directory; manuscript inputs and outputs are not used.

```python
from pathlib import Path
import sys
import json
import numpy as np
import pandas as pd
import scgeo as sg
from IPython.display import display

cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents)
             if (p / "scripts/scgeo_v020_examples/SOURCE_PART1.json").is_file()), None)
if root is None:
    raise RuntimeError("Open this notebook from the scgeo-notebooks checkout.")
examples = root / "scripts/scgeo_v020_examples"
if str(examples) not in sys.path:
    sys.path.insert(0, str(examples))  # only tutorial helpers, not package source
from tutorial_support import setup, save_preview, show_gallery, FORMATS
out = setup(root, "04_velocity_fields")
```

```python
from velocity_field_gallery import make_velocity_demo, run_gallery
adata = make_velocity_demo(seed=71)
result = sg.pl.velocity_field(adata, basis="display", velocity_basis="display",
    condition_key="condition", groups=["control", "treated"], mode="grid",
    background="selected", grid_size=25, n_neighbors=30, min_neighbors=6,
    min_coherence=0.2, return_result=True, show=False)
save_preview(result, out, "local_grid")
display(result.data.keys())
```

## Compare at common grid points

Use one coordinate extent and common scaling for the two selected conditions. Unsupported local directions remain unavailable.

```python
result = sg.pl.velocity_condition_compare(adata, condition_key="condition",
    group0="control", group1="treated", basis="display", mode="grid",
    grid_size=25, n_neighbors=30, min_neighbors=6, min_coherence=.2,
    scale="auto", return_result=True, show=False)
save_preview(result, out, "matched_conditions")
```
