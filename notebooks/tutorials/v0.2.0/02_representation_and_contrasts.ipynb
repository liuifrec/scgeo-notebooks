# 02 · Representation stability and evidence comparisons

**ScGeo 0.2.0 learning track.** Use related rotated coordinates and three simulated conditions to illustrate representation sensitivity, multiple contrasts and evidence availability.

Use the [separate tutorial environment](README.md). Run cells in order in a fresh kernel. Outputs are written to a new ignored results directory; manuscript inputs and outputs are not used.

```python
from pathlib import Path
import sys
import json
import numpy as np
import pandas as pd
import scgeo as sg
from IPython.display import display

cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents)
             if (p / "scripts/scgeo_v020_examples/SOURCE_PART1.json").is_file()), None)
if root is None:
    raise RuntimeError("Open this notebook from the scgeo-notebooks checkout.")
examples = root / "scripts/scgeo_v020_examples"
if str(examples) not in sys.path:
    sys.path.insert(0, str(examples))  # only tutorial helpers, not package source
from tutorial_support import setup, save_preview, show_gallery, FORMATS
out = setup(root, "02_representation_and_contrasts")
```

```python
from plotting_gallery import make_gallery_data, run_gallery
adata = make_gallery_data(seed=17)
sg.tl.robust_shift(adata, rep="X_toy", by="state", condition_key="condition",
                   group0="control", group1="treated", sample_key="sample",
                   center="geometric_median", n_boot=40, seed=17)
sg.tl.representation_stability(adata, reps=["X_toy", "X_rotated"], node_key="state",
    condition_key="condition", group0="control", group1="treated", sample_key="sample",
    center="geometric_median", velocity_keys={"X_toy":"V_toy", "X_rotated":"V_rotated"},
    n_boot=12, min_cells=10, seed=17)
report = sg.get.state_report(adata, node_key="state")
display(report)
result = sg.pl.representation_stability_heatmap(adata, metric="alignment_cosine",
    state_order=["State A", "State B", "State C"], rep_order=["X_toy", "X_rotated"],
    return_result=True, show=False)
save_preview(result, out, "representation_alignment")
```
