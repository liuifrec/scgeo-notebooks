# 03 · PAGA shifts, stored alignment and graph focus

**ScGeo 0.2.0 learning track.** A branched simulated state graph illustrates display arrows, pooled composition and stored quantitative alignment. The supplied symmetric graph is constructed, not inferred topology.

Use the [separate tutorial environment](README.md). Run cells in order in a fresh kernel. Outputs are written to a new ignored results directory; manuscript inputs and outputs are not used.

```python
from pathlib import Path
import sys
import json
import numpy as np
import pandas as pd
import scgeo as sg
from IPython.display import display

cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents)
             if (p / "scripts/scgeo_v020_examples/SOURCE_PART1.json").is_file()), None)
if root is None:
    raise RuntimeError("Open this notebook from the scgeo-notebooks checkout.")
examples = root / "scripts/scgeo_v020_examples"
if str(examples) not in sys.path:
    sys.path.insert(0, str(examples))  # only tutorial helpers, not package source
from tutorial_support import setup, save_preview, show_gallery, FORMATS
out = setup(root, "03_paga_shift_maps")
```

```python
from paga_shift_map_gallery import make_paga_demo, compute_demo_evidence, STATES, run_gallery
adata = compute_demo_evidence(make_paga_demo(seed=41), seed=41)
display(adata.obs.groupby(["state", "condition"], observed=True).size().unstack(fill_value=0))
common = dict(node_key="state", condition_key="condition", group0="control",
              group1="treated", basis="display", paga_key="paga_fixture",
              graph_node_order=STATES, background="contrast", min_cells=15,
              return_result=True, show=False)
result = sg.pl.paga_shift_map(adata, **common, node_color_mode="alignment",
    alignment_source="stored", analysis_rep="X_synthetic", vmin=-1, vmax=1,
    focus_states=["Upper late", "Lower early"], edge_focus="incident")
save_preview(result, out, "stored_alignment_focus")
display(result.data.keys())
```
