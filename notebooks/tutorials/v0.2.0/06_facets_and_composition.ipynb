# 06 · Matched embedding facets and figure composition

**ScGeo 0.2.0 learning track.** Compare annotated populations, scores and projected vectors on the same display embedding. The fixture names its synthetic coordinates X_umap for API compatibility; they are not fitted UMAP or inferred trajectories.

Use the [separate tutorial environment](README.md). Run cells in order in a fresh kernel. Outputs are written to a new ignored results directory; manuscript inputs and outputs are not used.

```python
from pathlib import Path
import sys
import json
import numpy as np
import pandas as pd
import scgeo as sg
from IPython.display import display

cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents)
             if (p / "scripts/scgeo_v020_examples/SOURCE_PART1.json").is_file()), None)
if root is None:
    raise RuntimeError("Open this notebook from the scgeo-notebooks checkout.")
examples = root / "scripts/scgeo_v020_examples"
if str(examples) not in sys.path:
    sys.path.insert(0, str(examples))  # only tutorial helpers, not package source
from tutorial_support import setup, save_preview, show_gallery, FORMATS
out = setup(root, "06_facets_and_composition")
```

```python
from embedding_facets_gallery import synthetic, run as facets_gallery
from plotting_composition_gallery import run as composed_gallery
adata = synthetic(seed=31)
result = sg.pl.embedding_facets(
    adata, facet="condition", color="state", velocity="grid",
    shared_limits=True, shared_velocity_scale=True, context="all",
    velocity_options={"grid_size":22, "n_neighbors":24, "min_neighbors":5},
    return_result=True, show=False)
display(result.data["facets"])
save_preview(result, out, "matched_facets")
```

## Facet by two annotations

Keep limits and color scales shared. The drawing budget is per facet, and does not redefine the numerical populations.

```python
result = sg.pl.embedding_facets(
    adata, facet_row="batch", facet_col="condition", color="illustrative_score",
    shared_limits=True, max_points=100, random_state=17,
    return_result=True, show=False)
save_preview(result, out, "two_way_score_facets")
```
