# GSE280305 — native plotting results

Browse 11 unchanged ScGeo 0.2.1 native SVG exports. See [the gallery and audit](README.md) and [the complete native application](../../v0.2.0/08_gse280305_native_visualization.ipynb).

The comparison is descriptive: D8 is post-irradiation; verified biological replicates and original projected velocity are unavailable. No new plot renderer is used here.

```python
from pathlib import Path
import csv
import json
import hashlib
from IPython.display import SVG, display

cwd = Path.cwd().resolve()
candidates = (cwd, cwd / "notebooks/applications/results/gse280305")
root = next((x for x in candidates if (x / "native_plot_manifest.json").is_file()), None)
if root is None:
    raise RuntimeError("Run from the gallery directory or companion repository root.")
def sha(path):
    return hashlib.sha256(path.read_bytes()).hexdigest()

manifest_path = root / "native_plot_manifest.json"
assert sha(manifest_path) == "310b761ec78b84a3430e12885c4f38f68e4140ff04a35786108456ccbba0a664"
manifest = json.loads(manifest_path.read_text())
assert manifest["native_plot_count"] == 11
assert manifest["scgeo_vcs_commit"] == "683228b1c52808d3cc5df2de5f6e104932c29328"
assert manifest["inference_scope"] == "descriptive_only"
for name, info in manifest["plots"].items():
    for path in (root / "figures" / info["figures"]["svg"],
                 root / "metadata" / info["metadata"]):
        assert sha(path) == manifest["file_sha256"][path.name], path
assert sha(root / "state_report.csv") == manifest["file_sha256"]["state_report.csv"]
with (root / "state_report.csv").open(newline="") as f:
    states = list(csv.DictReader(f))
assert len(states) == 14 and len({row["state"] for row in states}) == 14
assert all(row["inference_scope"] == "descriptive_only" for row in states)
print("PASS: 11 original SVGs, 11 original plot JSONs and 14 state-level rows.")
print("ScGeo commit:", manifest["scgeo_vcs_commit"])
```
