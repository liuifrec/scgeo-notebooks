# GSE249479 native ScGeo outputs
Review unchanged native figure exports without installing ScGeo or loading raw data. All figures retain their original analysis scope; this viewer does not recalculate or redraw them.

```python
from pathlib import Path
import hashlib, json
from IPython.display import display, SVG, Image, Markdown
here = Path.cwd()
if not (here / "gallery_manifest.json").is_file():
    here = here / "notebooks/applications/results/gse249479/native"
m = json.loads((here / "gallery_manifest.json").read_text())
for item in m["selected"]:
    figure = here / item["figure"]
    metadata = here / item["metadata"]
    assert hashlib.sha256(figure.read_bytes()).hexdigest() == item["figure_sha256"]
    assert hashlib.sha256(metadata.read_bytes()).hexdigest() == item["metadata_sha256"]
    display(Markdown("### " + item["name"] + "\n\n`scgeo.pl." + item["api"] + "`"))
    display(SVG(filename=str(figure)) if figure.suffix == ".svg" else Image(filename=str(figure)))
print("Verified", len(m["selected"]), "unchanged native figures")
```
